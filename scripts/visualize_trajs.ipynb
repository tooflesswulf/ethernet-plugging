### Visualize the plugin
1. Skip pickup and unplug.
2. Plot x-y-z

```python
import os 
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
def get_dataset(path):
    loaded =np.load(path)
    states, grippers, lens = loaded['pose'], loaded['gripper_width'], loaded['traj_length']
    ep_states, ep_grippers = [], []
    start = 0
    for l in lens:
        end = start + l
        ep_states.append(states[start:end])
        ep_grippers.append(grippers[start:end])
        start = end 
    return ep_states, ep_grippers 
def get_segment(pose, grippers, thr=18):
    pose3d = pose[:, :3]
    indices = np.arange(len(pose3d))
    closed_grippers = grippers < thr
    chosen_pose3d =  pose3d[closed_grippers]; chosen_indices = indices[closed_grippers]
    chosen_indices -= chosen_indices[0]
    # get the first segment thatt has closed grippers, ignore the reset
    indices = np.arange(len(chosen_indices))
    contiguous_indices = chosen_indices [chosen_indices == indices  ]
    return chosen_pose3d[contiguous_indices]
```

```python
def set_axes_equal(ax):
    x_limits = ax.get_xlim3d()
    y_limits = ax.get_ylim3d()
    z_limits = ax.get_zlim3d()

    x_range = abs(x_limits[1] - x_limits[0])
    y_range = abs(y_limits[1] - y_limits[0])
    z_range = abs(z_limits[1] - z_limits[0])

    radius = 0.5 * max(x_range, y_range, z_range)

    x_middle = np.mean(x_limits)
    y_middle = np.mean(y_limits)
    z_middle = np.mean(z_limits)

    ax.set_xlim3d([x_middle - radius, x_middle + radius])
    ax.set_ylim3d([y_middle - radius, y_middle + radius])
    ax.set_zlim3d([z_middle - radius, z_middle + radius])
```

```python
dir_path = '/zfsauton/scratch/yiqiw2/100%/datasets'
bc_path = os.path.join(dir_path, 'ethernet_plugin_unplug_50dataset', 'states.npz')
exbc_path = os.path.join(dir_path, 'ethernet_plugin_unplug_dataset', 'states.npz')
rl_path = os.path.join(dir_path, 'ethernet_plugin_unplug_rl_dataset', 'states.npz')
bc_poses, bc_grippers = get_dataset(bc_path)
rl_poses, rl_grippers = get_dataset(rl_path)
exbc_poses, exbc_grippers = get_dataset(exbc_path)
exbc_poses = exbc_poses[-25:]; exbc_grippers = exbc_grippers[-25:]
```

```python
bc_segs = [ get_segment(pose, gripper) for (pose, gripper) in zip(bc_poses, bc_grippers) ]
rl_segs = [ get_segment(pose, gripper) for (pose, gripper) in zip(rl_poses, rl_grippers) ]
exbc_segs = [ get_segment(pose, gripper) for (pose, gripper) in zip(exbc_poses , exbc_grippers) ]
```

```python

fig = go.Figure()

# BC trajectories
for i, traj in enumerate(bc_segs):
    fig.add_trace(
        go.Scatter3d(
            x=traj[:, 0],
            y=traj[:, 1],
            z=traj[:, 2],
            mode="lines",
            name="BC",
            legendgroup="BC",
            showlegend=(i == 0),  # Only first BC trajectory appears in legend
            line=dict(color="blue"),
        )
    )

# RL trajectories
for i, traj in enumerate(rl_segs):
    fig.add_trace(
        go.Scatter3d(
            x=traj[:, 0],
            y=traj[:, 1],
            z=traj[:, 2],
            mode="lines",
            name="RL",
            legendgroup="RL",
            showlegend=(i == 0),  # Only first RL trajectory appears in legend
            line=dict(color="red"),
        )
    )

fig.update_layout(
    scene=dict(aspectmode="data"),
    legend=dict(itemsizing="constant"),
)

fig.show()
```
